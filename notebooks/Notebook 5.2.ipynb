```python
import time
 
LANDING = "/Volumes/nyc_taxi/bronze/streaming_landing"
 
sample = spark.table("nyc_taxi.silver.trips_clean").limit(20000)
batches = sample.randomSplit([1.0] * 20, seed=42)
 
for i, batch in enumerate(batches):
    (batch.write.format("json").mode("overwrite")
        .save(f"{LANDING}/batch_{i:03d}"))
    print(f"landed batch {i:03d}: {batch.count()} events")
    time.sleep(15)
```
