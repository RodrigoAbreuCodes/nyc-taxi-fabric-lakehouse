```sql
WITH weekly_revenue AS (
  SELECT
    date_trunc('week', date_key) AS week_start,
    COUNT(DISTINCT date_key)     AS days_in_week,
    SUM(total_amount)            AS revenue
  FROM nyc_taxi.gold.fact_trips
  GROUP BY date_trunc('week', date_key)
)
SELECT
  date_format(week_start, 'yyyy-MM') AS month,
  month(week_start)                  AS month_num,
  ROUND(AVG(revenue), 2)             AS avg_weekly_revenue,
  COUNT(*)                           AS full_weeks_in_month
FROM weekly_revenue
WHERE days_in_week = 7
GROUP BY date_format(week_start, 'yyyy-MM'), month(week_start)
ORDER BY month_num;
```
